```python
import pandas as pd
import seaborn as sns
import numpy as np
from matplotlib import pyplot as plt
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split, cross_val_score

from sklearn.preprocessing import LabelEncoder,StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
columns_with_null_cat_none = [
    'Alley', 'FireplaceQu', 'PoolQC', 'Fence', 'MiscFeature',
    'MasVnrType', 'BsmtQual', 'BsmtCond', 'BsmtExposure',
    'BsmtFinType1', 'BsmtFinType2','GarageType',
    'GarageFinish', 'GarageQual', 'GarageCond'
]

columns_with_null_cat_mode = [
    'MSZoning', 'Utilities', 'Exterior1st', 'Exterior2nd',
    'KitchenQual', 'Functional', 'SaleType', 'Electrical'
]

columns_with_null_num_zero = [
    'MasVnrArea', 'GarageYrBlt', 'BsmtFinSF1', 'BsmtFinSF2', 'BsmtUnfSF',
    'TotalBsmtSF', 'BsmtFullBath', 'BsmtHalfBath', 'GarageCars', 'GarageArea'
]

columns_with_null_num_median = [
    'LotFrontage'
]
```

```python
for col in columns_with_null_cat_none:
    df[col].fillna('None', inplace=True)
    dt[col].fillna('None', inplace=True)
```

```python
for col in columns_with_null_cat_mode:
    df[col].fillna(df[col].mode()[0], inplace=True)
    dt[col].fillna(df[col].mode()[0], inplace=True)
```

```python
for col in columns_with_null_num_zero:
    df[col].fillna(0, inplace=True)
    dt[col].fillna(0, inplace=True)
```

```python
for col in columns_with_null_num_median:
    df[col].fillna(df[col].median(), inplace=True)
    dt[col].fillna(df[col].median(), inplace=True)
```

```python
df['Electrical'].fillna(df['Electrical'].mode()[0], inplace=True)
dt['Electrical'].fillna(df['Electrical'].mode()[0], inplace=True)
```

```python
#df.isnull().sum().sort_values(ascending=False).head(35)
```

```python
#dt.isnull().sum().sort_values(ascending=False).head(35)
```

```python
df["SaleCondition"] = [1 if each == "Normal" else 0 for each in df["SaleCondition"]]
dt["SaleCondition"] = [1 if each == "Normal" else 0 for each in dt["SaleCondition"]]
```

```python
le = LabelEncoder()
```

```python
df_object = df.select_dtypes('object')
dt_object = dt.select_dtypes('object')
df_object.head()
```

```python
for col in df_object:
    df[col] = le.fit_transform(df[col])
for col in dt_object:
    dt[col] = le.fit_transform(dt[col])
```

```python
#df['Street'].value_counts()
```

```python
score_list = {}
```

## 1. Train a KNN Regressor

```python
X = df.drop("SalePrice",axis=1)
y = df["SalePrice"]

X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=1,test_size=0.3)
```

```python
knn = KNeighborsRegressor(n_neighbors=10)

knn.fit(X_train, y_train)
knn_y_pred = knn.predict(X_test)
```

- Perform cross validation

```python
scores = cross_val_score(knn, X, y, cv=10)
print("%0.2f accuracy with a standard deviation of %0.2f" % (scores.mean(), scores.std()))


knn_score = knn.score(X_test,y_test)
score_list["KNN Classifier"] = knn_score
```

## 2. Train a SVM Regression

```python
svr = SVR(kernel="rbf", C=100, epsilon=0.2)
svr.fit(X_train, y_train)
svr_y_pred = svr.predict(X_test)
```

- Perform cross validation

```python
scores = cross_val_score(svr, X, y, cv=4)

scores
print("%0.2f accuracy with a standard deviation of %0.2f" % (scores.mean(), scores.std()))


svr_score = svr.score(X_test,y_test)
score_list["SVR"] = svr_score
```

## 3. Train a Decision Tree Regression

```python
dtr = DecisionTreeRegressor(random_state=1)
dtr.fit(X_train,y_train)
dtr_y_pred = dtr.predict(X_test)
```

- Perform cross validation

```python
scores = cross_val_score(dtr, X, y, cv=4)

scores
print("%0.2f accuracy with a standard deviation of %0.2f" % (scores.mean(), scores.std()))


dtc_score = dtr.score(X_test,y_test)
score_list["DTC"] = dtc_score
```

## 4. Train a Random Forest Regression

```python
rfr = RandomForestRegressor(n_estimators=10,random_state=1)
rfr.fit(X_train,y_train)
rfr_y_pred = rfr.predict(X_test)
```

```python
rfr_score = rfr.score(X_test,y_test)
print(rfr_score)
#score_list["RFR"] = rfr_score
```

## 5. Compare all the performance of all regression models
